```python
import pandas as pd
import requests
from google.colab import userdata

API_KEY = userdata.get('API_KEY')

headers = {
    'Authorization': f'Bearer {API_KEY}'
}

link_api = "https://apidadosabertos.saude.gov.br/vigilancia-e-meio-ambiente/mpox?limit=100&offset=0"

parametros = {
    'sg_uf': 'SC',
    'id_mn_resi': '4201406'
}

resposta = requests.get(link_api, params=parametros, headers=headers)
print(f'Status: {resposta.status_code}')

if resposta.status_code == 200:
    dados = resposta.json()
    df = pd.DataFrame(dados['mpox'])

    # estrutura básica

    print(df.shape)

    # tipos das variáveis
    print(df.dtypes)

    # percentual de nulos por coluna
    print('\nNULOS POR COLUNA (%)')
    print(df.isnull().mean().mul(100).round(2).sort_values(ascending=False))

    # duplicatas
    print(f'Registros duplicados: {df.duplicated().sum()}')

    # valores únicos por coluna
    print(df.nunique().sort_values())

    # estatísticas das colunas numéricas
    print(df.describe())

    # primeiros registros
    display(df.head())

    # limpeza
    df.columns = df.columns.str.strip().str.upper()
    df = df.dropna(how='all')
    df = df.drop_duplicates()
    df = df.fillna('SEM DADO')
    df = df.reset_index(drop=True)
```

A estrutura básica em 5 linhas e 54 colunas, e, o tipo original das maioria das variáveis é object. Pode se notar também o fato de alguns dados cruciais sendo ocultos, como a data de internação ("dt_interna") e a data de vacinação ("data_vacina"), que poderiam vir a ser importantes para uma análise mais precisa dos dados. Pelo analisado, também podemos perceber que, não houveram dados nesta API que não fazem sentido, pelo menos não na parte de "sg_uf: SC" e "id_mn_resi: 4201406", já que esses parametros ja foram postos na API para mostrar e filtrar esses dados. Na questão de duplicidade, também não houveram dados que não era pra terem sidos duplicados, porém há sim uma grande quantidade de dados duplicados em várias partes da API, como na parte da data da coleta("dt_coleta") e na data de notificação ("dt_notific"). E por último, dá pra perceber que a API do DATASUS usa sim um dicionário que precisaria ter uma tradução, assim como nos gêneros, código de raça, código da unidade federal e etc.
