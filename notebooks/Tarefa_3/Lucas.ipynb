```python
import requests
import pandas as pd
import time
from google.colab import userdata


api_key = userdata.get("API_KEY")

headers = {"accept": "text/csv","chave-api-dados": api_key}

lista_dfs = []

for pagina in range(0, 200):

    url = ("https://apidadosabertos.saude.gov.br/"f"arboviroses/zikavirus?limit=20&offset={pagina}")

    response = requests.get(url,headers=headers)

    print(f"Página {pagina} - Status:", response.status_code)

    try:

        df_temp = pd.read_csv(pd.io.common.StringIO(response.text),sep=";")

        lista_dfs.append(df_temp)

    except:

        print(f"Erro ao ler página {pagina}")

    time.sleep(0.2)


df = pd.concat(lista_dfs,ignore_index=True)


df.to_csv("zikavirus_completo.csv", sep=";",index=False)

print("\nBase final carregada.")

df.head()
```
