## Mortalidade Materna



```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
columns = ['TIPOBITO', 'IDADEMAE', 'ESCMAE2010', 'GRAVIDEZ', 'PARTO', 'OBITOPARTO', 
           'PESO','IDADE','OCUPMAE','QTDFILMORT',
           'QTDFILVIVO', 'LOCOCOR','ASSISTMED']

df = pd.read_csv('../data/processed/sim_sc_processado_analitico.csv')
df = df[columns]
df = df[df['IDADE'] < 10]
df.info()
print("Quantidade de óbitos: ", len(df))
```

1. Análise de Distribuição e Dispersão:
    Os gráficos devem permitir visualizar de forma clara a distribuição de variáveis contínuas cruciais, como o peso do bebê ao nascer e o tempo de gestação em semanas, facilitando a identificação de anomalias (extremos de baixo peso ou prematuridade extrema).

```python
sns.lineplot(data=df["IDADE"])
```

```python
sns.scatterplot(
    data=df, 
    x="GRAVIDEZ", 
    y="PESO", 
    hue="OBITOPARTO"
)

plt.title("Tipo de Gravidez e o Peso", fontsize=14, fontweight="bold")
plt.xlabel("Tipo de Gravidez")
plt.ylabel("Peso")

plt.show()
```

2. Fatores do Parto e Gravidez:
    Deve conseguir cruzar o momento e o tipo do óbito com o tipo de parto e o tipo de gravidez para entender o contexto clínico do evento.

```python
#Óbito em parto
sns.countplot(data=df, x="OBITOPARTO")

plt.title("Óbitos em parto", fontsize=14, fontweight="bold")
plt.xlabel("Óbito em Parto")
plt.ylabel("Quantidade")

plt.show()
```

```python
#Idades dos Falecidos
sns.countplot(data=df, x="IDADE")

plt.title("Idades", fontsize=14, fontweight="bold")
plt.xlabel("Idades")
plt.ylabel("Quantidade")

plt.show()
```

```python
#Gravidez e Tipo do óbito
sns.scatterplot(data=df, x="GRAVIDEZ", y="TIPOBITO")

plt.title("Relação entre gravidez e o tipo de óbito", fontsize=14, fontweight="bold")
plt.xlabel("Gravidez")
plt.ylabel("Tipo de Óbito")

plt.show()
```

```python
#Gravidez e Parto
sns.scatterplot(data=df, x="GRAVIDEZ", y="PARTO")

plt.title("Relação entre gravidez e o tipo de parto realizado", fontsize=14, fontweight="bold")
plt.xlabel("Gravidez")
plt.ylabel("Tipo de Parto")

plt.show()
```

3. Perfil Socioeconômico e Demográfico:
    A solução precisa evidenciar o perfil da mãe, permitindo segmentar os dados pela idade materna e pelo nível de escolaridade para identificar vulnerabilidades sociais.

```python
#Idade da Mãe
sns.countplot(data=df, x="IDADEMAE")

plt.title("Idade da Mãe", fontsize=14, fontweight="bold")
plt.xlabel("Faixas Etária")
plt.ylabel("Quantidade")

plt.show()
```

```python
#Escolaridade da Mãe
sns.countplot(data=df, x="ESCMAE2010")

plt.title("Escolaridade da Mãe", fontsize=14, fontweight="bold")
plt.xlabel("Escolaridade Mãe 2010")
plt.ylabel("Quantidade")

plt.show()
```

```python
#Ocupação da Mãe
sns.countplot(data=df, x="OCUPMAE")

plt.title("Ocupação da Mãe", fontsize=14, fontweight="bold")
plt.xlabel("Ocupação")
plt.ylabel("Quantidade")

plt.show()
```

4. Período e Circunstâncias do Óbito:
    Deve ficar claro na análise quando a morte ocorreu (se foi durante o parto, no período fetal, se foi um óbito materno na gestação/puerpério, etc.), trazendo luz à janela de tempo mais crítica para a intervenção em saúde.

```python
#Relação entre o tipo de morte e a assitência médica prestada
sns.scatterplot(data=df, x="TIPOBITO", y="ASSISTMED")

plt.title("Relação entre o tipo de óbito e a assistência médica", fontsize=14, fontweight="bold")
plt.xlabel("Tipo de Óbito")
plt.ylabel("Assistência Médica")

plt.show()
```
