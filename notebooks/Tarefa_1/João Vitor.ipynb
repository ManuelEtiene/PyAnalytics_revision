```python
import requests


api_key =  ""
link_api = "https://apidadosabertos.saude.gov.br/assistencia-a-saude/hospitais-e-leitos?limit=100&offset=0"
parametros = {
    "key": api_key,
    "nome_da_regiao_do_brasil_onde_fica_o_hospital": "SUL"

}
resposta = requests.get(link_api, params=parametros)

print(resposta)
print(resposta.content)
if resposta.status_code == 200:
    dados_requisicao = resposta.json()
    print(dados_requisicao)

```

```python
import pandas as pd

estabelecimentos = dados_requisicao["hospitais_leitos"]
df_estabelecimentos = pd.DataFrame(estabelecimentos)
display(df_estabelecimentos.head())
```

```python
df_sul = df_estabelecimentos[df_estabelecimentos['nome_da_regiao_do_brasil_onde_fica_o_hospital'] == 'SUL']

display(df_sul.head())
```

```python
import matplotlib.pyplot as plt


contagem_tipos = df_sul['tipo_da_gestao_do_hospital'].value_counts().sort_index()

plt.bar(contagem_tipos.index, contagem_tipos.values)

plt.title('Número de hospitais por tipo de gestão na região Sul (Dupla = D, Estadual = E, Municipal = M)')
plt.xlabel('Tipo de Gestão')
plt.ylabel('Número de Hospitais')

plt.show()
```
