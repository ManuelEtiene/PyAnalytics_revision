```python
import pandas as pd

with pdfplumber.open('remume_2024.pdf') as pdf:
    for pagina in pdf.pages:
        tabelas = pagina.extract_tables()
        for tabela in tabelas:
            if tabela:
                df = pd.DataFrame(tabela[1:], columns=tabela[0])
                df.dropna(how='all', inplace=True)
                display(df)
```
