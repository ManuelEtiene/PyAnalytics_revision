```python
file_id = '1YKdnI4ziFCq9hoeXqTaje3BIL7BVwyI4'
url = f'https://drive.google.com/uc?id={file_id}'
output = 'remume_versao_2024.pdf'
gd.download(url,output,quiet=False) #Download do arquivo do Remume
```

```python
pdf_path = "remume_versao_2024.pdf"

texto_completo = ""

with pdfplumber.open(pdf_path) as pdf:
    for pagina in pdf.pages:
        texto = pagina.extract_text()

        if texto:
            texto_completo += texto + "\n"

print(texto_completo) # feito para mostrar tudo escrito no PDF
```

```python
tabelas_extraidas = []

with pdfplumber.open(pdf_path) as pdf:
    for i, pagina in enumerate(pdf.pages):

        tabelas = pagina.extract_tables()

        for tabela in tabelas:
            df = pd.DataFrame(tabela)

            tabelas_extraidas.append(df)

            print(f"\nTabela encontrada na página {i+1}") # mostra todas as tabelas encontradas no PDF
```

```python
for i, tabela in enumerate(tabelas_extraidas):
    tabela.to_csv(f"tabela_{i+1}.csv", index=False) #cria um CSV de cada tabela encontrada e enumera elas
```
