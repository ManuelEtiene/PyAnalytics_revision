```python
import pdfplumber
import pandas as pd

arquivo_pdf = "remume_versao_2024.pdf"

#Extração do Texto

texto_completo = ""

with pdfplumber.open(arquivo_pdf) as pdf:

    total_paginas = len(pdf.pages)

    print(f"Total de páginas: {total_paginas}")

    for i, pagina in enumerate(pdf.pages):

        texto = pagina.extract_text()

        if texto:

            texto_completo += texto + "\n"

        print(f"Página {i+1} processada")

print("\nPrimeiros caracteres extraídos:\n")

print(texto_completo[:3000])
```

```python
# Salvar texto

with open("texto_remume.txt", "w", encoding="utf-8") as arquivo:

    arquivo.write(texto_completo)
```
