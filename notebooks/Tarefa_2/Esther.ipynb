##Passo 2: Extração de Dados de PDFs e o Desafio "Remume"

```python
!pip install pdfplumber pandas -q

import pdfplumber
import pandas as pd

# Abri o arquivo PDF da REMUME
pdf_path = "remume_versao_2024.pdf"
dados_completos = []

with pdfplumber.open(pdf_path) as pdf:
    # Percorri as 5 primeiras páginas para extrair as tabelas de medicamentos
    for i, pagina in enumerate(pdf.pages[:5]):
        tabelas = pagina.extract_tables()

        for tabela in tabelas:
            for linha in tabela:
                # Armazenei cada linha da tabela de medicamentos
                dados_completos.append(linha)

# Converti os dados extraídos em um DataFrame do Pandas para estruturar a lista
df_remume = pd.DataFrame(dados_completos)

# Visualizei as 10 primeiras linhas da tabela extraída
print("Visualização dos medicamentos extraídos:")
display(df_remume.head(10))

# Salvei o resultado final em um arquivo CSV
df_remume.to_csv("remume_extraido.csv", index=False)
print("\nDesafio concluído: Dados extraídos e salvos em 'remume_extraido.csv'")
```
